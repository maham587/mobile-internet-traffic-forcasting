**Generate one CSV file from the 62 txt files**

* The telecommunications data for the area of milano for this work is organized in 62 txt file for 2013-11-01 to 2014-01-01 period, with each txt files around 1 GB each.
* In order to ease data processing, homogenity and retrival for the duration of the two months and  to proceed dvanced Filtering with dataset, we organized the 62 txt files in one pandas Data frame with only relevant columns for this task and save it in csv file to be loaded once.

```python
# Import Libraries
import pandas as pd
```

```python

files = ['dataset1/sms-call-internet-mi-2013-11-%02d.txt'%i for i in range (1, 31)] + ['dataset1/sms-call-internet-mi-2013-12-%02d.txt'%i for i in range (1, 32)] + ['dataset1/sms-call-internet-mi-2014-01-01.txt']

frames = [pd.read_csv(file, sep=None, engine='python', header= None) for file in files]
df_cdrs = pd.concat(frames, ignore_index=True)


columns = ["Square id", "Time Interval", "Internet traffic activity"] # relevant columns name for this work
cols = [2,3,4,5,6] # columns index to be drop about sms and call

# Reduce columns to only relevant to this study
df_cdrs.drop(df_cdrs.columns[cols],axis=1,inplace=True)
df_cdrs.columns = columns


# save a csv file to be load once
df_cdrs = df_cdrs.fillna(0)   
df_cdrs.to_csv('internet_traffic_milano_nov-dec-2013.csv', index=False)
```
